### What is a database?

- Organized collection of structured information of data, which is stored on a computer system.

- The data can be easily accessed

- The data can be modified

- The data can be controlled and organized

- Many databases use a structured query language (SQL) to modify and write data

### What is data?

- Data can be related to just about any object.

- For example, a user on an application may have:
    - Name
    - Age
    - Email
    - Password

- All of this is data

- A database is a collection of data.

- Since data, on its own, is just data. A database allows management of data.

- Databases are organized in how data can be retrieved, stored and modified.

- There are many types of Database Management Systems (DBMS) eg. SQLite, Postgres

## What is SQL?

- Pronounced either as S-Q-L or "See Quel"

- Standard language for dealing with dealing with relational databases.

- SQL can be used to do different things with databases like:
    - Create
    - Read
    - Update
    - Delete
