## 1. What is FastAPI?

FastAPI is a modern Python web framework for building **RESTful APIs**.

The name has two meanings of "Fast":
- **Fast (Performance):** One of the fastest Python frameworks — rivals NodeJS and Go
- **Fast (Development):** Auto docs, type hints, and data validation out of the box

### Key Features
| Feature | What it means |
|---------|---------------|
| Few Bugs | Type hints catch errors before runtime |
| Quick & Easy | Less boilerplate than Flask/Django |
| Robust | Production-ready, used by Netflix, Uber, Microsoft |
| Standards | Built on OpenAPI and JSON Schema standards |

### 💡 Why this matters for AI roles
FastAPI is the **industry standard for serving ML models and LLM APIs**.  
Every AI backend you'll build will likely use FastAPI to expose predictions as REST endpoints.

---
## 2. Request & Response Cycle

```
Client (Browser/App)
        │
        │  HTTP Request (GET, POST, PUT, DELETE)
        ▼
  FastAPI Server
        │
        │  HTTP Response (JSON data)
        ▼
Client (Browser/App)
```

- The **client** sends a request to FastAPI
- FastAPI **processes** the request (reads DB, runs ML model, etc.)
- FastAPI sends back a **response** — usually JSON

FastAPI also ships with a built-in **Swagger UI** at `/docs` — a free interactive page to test your API without needing Postman.

## 3. HTTP Request Methods (CRUD)

| CRUD Operation | HTTP Method | Example Use Case |
|----------------|-------------|------------------|
| **Create** | `POST` | Add a new book to the database |
| **Read** | `GET` | Fetch all books / fetch one book by ID |
| **Update** | `PUT` | Edit a book's details |
| **Delete** | `DELETE` | Remove a book |

### 💡 Insight
Think of HTTP methods like database operations.  
Every API you ever build maps to one of these 4 actions.

## 4. FastAPI vs Flask — Key Differences

| Feature | Flask | FastAPI |
|---------|-------|---------|
| Speed | Slower | Much faster (async) |
| Auto docs | ❌ Manual | ✅ Auto Swagger UI |
| Data validation | ❌ Manual | ✅ Built-in via Pydantic |
| Type hints | Optional | Core feature |
| Best for | Simple web apps | APIs, ML serving |

### 💡 Insight
Flask is like a bicycle — simple and flexible.  
FastAPI is like a sports car — more powerful, built for speed and production AI systems.

### What is uvicorn?
- uvicorn is the **server** that runs your FastAPI app
- FastAPI = the framework (defines your routes)
- uvicorn = the engine (actually serves requests)
- Think of it like: FastAPI writes the menu, uvicorn runs the restaurant

## What are Status Codes?

- An HTTP status code is used to help the client (the user or system submitting data to the server) to understand what happened on the server side application.

- Status codes are international standards on how a client/server should handle the result of a request.

- It allows everyone who sends a request to know if their submission was successful or not.

## Status Codes:


- `1xx` --->  `Information Response: Request Processing`
- `2xx` --->  `Success: Request Successfully completed`
- `3xx` --->  `Redirection: Further action must be completed`
- `4xx` --->  `Client Errors: An error was caused by the client`
- `5xx` --->  `Server Errors: An error occurred on the server`


## 2xx Successful Status Codes:

- `200: OK` : standard response for a successful request. Commonly used for successful GET requests when data is being returned.

- `201: Created`: the request has been successful, creating a new resource. used when a POST creates an entity.

- `204: No Content`: the request has been successful, did not create an entity nor return anthing. Commonly used with PUT requests.

## 4xx Client Errors Status Codes

- `400: Bad Request`: Cannot process request due to client error. Commonly used for invalid request methods

- `401: Unauthorized`: Client does not have valid authentication for target resource

- `404: Not Found`: The client’s requested resource cannot be found

- `422: Unprocessable Entity`: Semantic errors in client request



## 5xx Server Status Codes

- `500: Internal Server Error`: Generic Error Message, when an unexpected issue on the server side happened.
